Tipos de datos y estructuras


Listas

Las estructuras de datos permiten almacenar varios valores en una sola variable.

Las más utilizadas en Python son:

- Listas : Ordenadas y modificables.  
- Diccionarios : Clave–valor, para representar información estructurada.  
- Tuplas : Ordenadas pero inmutables, ideales para datos que no deben cambiar.

```python
numeros = [10, 20, 30]
numeros.append(40)
print(numeros)
print("Primer elemento:", numeros[0])

# operaciones con listas
numeros = [10, 5, 3, 20]

# ordenar la lista
numeros.sort()
print("Lista ordenada:", numeros)

# insertar en una posición
numeros.insert(1, 99)
print("Insertar en posición 1:", numeros)

# eliminar un elemento
numeros.remove(99)
print("Eliminar 99:", numeros)

# largo de la lista
print("Cantidad de elementos:", len(numeros))
```

Diccionarios

```python
estudiante = {"nombre": "Gonzalo", "edad": 22, "promedio": 8.5}
print(estudiante["nombre"])

estudiante["edad"] = 23
print(estudiante)

persona = {"nombre": "Ana", "edad": 25, "ciudad": "La Plata"}

# recorrer
for clave, valor in persona.items():
    print(clave, ":", valor)

# eliminar una clave
persona.pop("ciudad")
print("Después de eliminar ciudad:", persona)
```

Tuplas

```python
# tupla con datos
coordenada = (10, 20)
print("Coordenada:", coordenada)
print("X:", coordenada[0])
print("Y:", coordenada[1])

# largo
print("Cantidad de elementos en la tupla:", len(coordenada))

# recorrer
print("Recorriendo la tupla:")
for valor in coordenada:
    print("Valor:", valor)

# ejemplo de uso con fecha
fecha_nacimiento = (4, 12, 2025)
dia, mes, año = fecha_nacimiento
print("Dia:", dia, "- Mes:", mes, "- Año:", año)

# pasaje de tupla a lista para modificarla
valores = (1, 2, 3)
lista_valores = list(valores)
lista_valores.append(4)
print("Lista modificada:", lista_valores)

# volver a tupla
tupla_nueva = tuple(lista_valores)
print("Tupla nueva:", tupla_nueva)
```

2) Funciones, parámetros y retorno

Una función es un bloque de código reutilizable.
Puede recibir parámetros y devolver un valor con return.

Sirven para:
- Organizar mejor el código
- Evitar repetir instrucciones

Funciones con retorno

```python
def suma(a, b):
    return a + b

resultado = suma(5, 7)
print("Resultado:", resultado)
```

Función más completa

```python
def evaluar_promedio(prom, minimo=6):
    if prom >= minimo:
        return "Aprobado"
    else:
        return "Desaprobado"

print(evaluar_promedio(8))
print(evaluar_promedio(5))
```

Programación Orientada a Objetos (POO)

La POO organiza el código en clases que representan entidades del mundo real.

Cada clase:

- tiene atributos : datos del objeto
- tiene métodos : acciones que puede realizar

Un objeto es una instancia de una clase.

Clase simple

```python
class Estudiante:
    def __init__(self, nombre):
        self.nombre = nombre
        self.notas = []

    def agregar_nota(self, nota):
        self.notas.append(nota)

    def promedio(self):
        return sum(self.notas) / len(self.notas)

    def estado(self):
        if self.promedio() >= 6:
            return "Aprobado"
        else:
            return "Desaprobado"

e = Estudiante("Gonzalo")
e.agregar_nota(9)
e.agregar_nota(6)
print("Estado del estudiante:", e.estado())
```
